## Plot Pearsons

```python
def PlotPearsons(base_dir, x_labels):
    """
    Plot violin plots of channel correlations averaged across all CSV files.
    """
    # Collect data from all CSV files
    all_data = []
    csv_files = sorted([f for f in os.listdir(base_dir) if f.endswith('.csv')])
    
    if len(csv_files) != len(x_labels):
        print(f"Warning: Found {len(csv_files)} CSV files but {len(x_labels)} labels provided")
    
    for i, csv_file in enumerate(csv_files):
        file_path = os.path.join(base_dir, csv_file)
        df = pd.read_csv(file_path)
        
        # Get the label for this file
        label = x_labels[i] if i < len(x_labels) else csv_file
        
        # Add channel 1 correlations
        for val in df['channel_1_correlation']:
            all_data.append({
                'file': label,
                'channel': 'Channel 1',
                'correlation': val
            })
        
        # Add channel 2 correlations
        for val in df['channel_2_correlation']:
            all_data.append({
                'file': label,
                'channel': 'Channel 2',
                'correlation': val
            })
    
    # Create DataFrame for plotting
    plot_df = pd.DataFrame(all_data)
    
    # Create violin plot
    fig, ax = plt.subplots(figsize=(9, 6))
    
    sns.violinplot(data=plot_df, x='file', y='correlation', hue='channel', 
                   split=False, ax=ax, palette=["#69bffd", "#a3f787"])
    
    # Add mean markers and values on the violins
    # Get unique files and channels for positioning
    files = plot_df['file'].unique()
    channels = plot_df['channel'].unique()
    
    # Calculate positions for each violin (seaborn places them automatically)
    for i, file in enumerate(files):
        for j, channel in enumerate(channels):
            # Filter data for this file and channel
            subset = plot_df[(plot_df['file'] == file) & (plot_df['channel'] == channel)]
            mean_val = subset['correlation'].mean()
            
            # Position: each file group has 2 violins (channels), offset by 0.2
            x_pos = i + (j - 0.5) * 0.4
            
            # Plot mean as a white diamond marker
            ax.plot(x_pos, mean_val, marker='D', color='white', 
                   markersize=8, markeredgecolor='black', markeredgewidth=1.5, 
                   zorder=3)
            
            # Add mean value as text
            ax.text(x_pos, mean_val, f'{mean_val:.3f}', 
                   ha='center', va='bottom', fontsize=8, 
                   fontweight='bold', color='black',
                   bbox=dict(boxstyle='round,pad=0.3', facecolor='white', 
                            edgecolor='black', alpha=0.8))
    
    ax.set_xlabel('Panel (Well)', fontsize=12)
    ax.set_ylabel('Pearson Correlation', fontsize=12)
    ax.set_title('Pearson Correlation between Predicted and Ground Truth', fontsize=16)
    ax.legend(title='Channel', fontsize=10)
    ax.grid(True, alpha=0.3, axis='y')
    
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    
    return fig, ax
```

```python
pearsons_dir = "test_results"
labels = ["B1", "B2", "B3"]

PlotPearsons(pearsons_dir, labels)
plt.show()
```
